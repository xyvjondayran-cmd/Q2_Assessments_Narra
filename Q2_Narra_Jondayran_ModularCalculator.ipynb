```python
# Make all the necessary functions for the calculator
def add_numbers(num1, num2):
  sum = num1 + num2
  return sum

def subtract_numbers(num1, num2):
  difference = num1 - num2
  return difference

def multiply_numbers(num1, num2):
  product = num1 * num2
  return product

def divide_numbers(num1, num2):
  quotient = num1 / num2
  return quotient

# Ask the user for num1 and num2
num1 = int(input("Enter first number num1:"))
num2 = int(input("Enter second number num2: "))

# Display the selected operation
print("\nChoose operation:")
print("1 - Addition")
print("2 - Subtraction")
print("3 - Multiplication")
print("4 - Division")

# Make the user choose an operation
choice = input("Enter choice: ")

# Call the correct function
if choice == "1":
    result = add_numbers(num1, num2)
elif choice == "2":
    result = subtract_numbers(num1, num2)
elif choice == "3":
    result = multiply_numbers(num1, num2)
elif choice == "4":
    if num2 == 0:
        result = "Cannot divide by zero."
    else:
        result = divide_numbers(num1, num2)
else:
    result = "Invalid choice."

# Display the result
print("Result:", result)
```

1. I created four functions: "add_numbers", "subtract_numbers", "multiply_numbers", and "divide_numbers".

2. Each function used the parameters "num1" and "num2".

3. The arguments passed were the two numbers entered by the user, "num1" and "num2".

4. The program stored the returned value in the "result" variable and displayed it to the user.

5. Dividing the program into functions makes the code more organized, easier to understand, and easier to reuse or modify.
